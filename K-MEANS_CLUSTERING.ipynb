#K-MEANS CLUSTERING
###IMPORTING MODULES

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np
import matplotlib.pyplot as plt
```

###DATASETS

```python
x=np.array([[1,2],[1,4],[1,0],[4,2],[4,4],[4,0],[10,2],[10,4],[10,0]])
```

###CREATING THE MODEL

```python
model=KMeans(n_clusters=3,random_state=42,n_init=10)
```

###FITTING THE MODEL

```python
model.fit(x)
label=model.labels_
```

###EVALUATING THE MODEL

```python
sil_score=silhouette_score(x,label)
print(f"silhouette score: {sil_score:.2f}")
```

###DATA VISUALIZATION

```python
plt.scatter(x[:,0],x[:,1],c=label,cmap='rainbow',marker='o',edgecolors='black')
plt.xlabel('x1')
plt.ylabel('x2')
plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],s=100,c="red",marker="x",label="centroids")
plt.title("K-Means Clustering")
plt.legend()
plt.show()
```
