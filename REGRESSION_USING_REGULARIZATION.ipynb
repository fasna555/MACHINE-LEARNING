#REGULARIZATION
###IMPORTING MODULES

```python
import numpy as np
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.preprocessing import PolynomialFeatures
import matplotlib.pyplot as plt
```

###CREATING DATASETS

```python
x=np.array([[12],[23],[34],[45],[56],[67],[78],[89],[123],[134]])
Y=np.array([240,1135,2568,4521,7865,9236,11932,14589,19856,23145])
```

###TRAINING THE MODEL

```python
model=RidgeCV()
poly=PolynomialFeatures(degree=3)
X=poly.fit_transform(x)
model.fit(X,Y)

```

###PREDICTING THE OUTPUT

```python
y_pred=model.predict(X)
print("coeff: ",model.coef_)
print("intercept: ",model.intercept_)
print("y predicted: ",y_pred)
```

###EVALUATING THE MODEL



```python
length=len(x)
R2=r2_score(Y,y_pred)
print("R2 score=",R2)
SSE=mean_squared_error(Y,y_pred)*length
print("SSE=",SSE)
```

###PLOTTING THE GRAPH

```python
plt.scatter(x,Y,color="red",label="data points")
plt.plot(x,y_pred,color="blue",label="regression curve")
plt.legend()
plt.xlabel("x")
plt.ylabel("y predicted")
plt.title("Regression using Regularization ")
plt.show()
```
