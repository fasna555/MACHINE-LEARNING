#POLYNOMIAL REGRESSION
###IMPORTING MODULES

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import PolynomialFeatures
import matplotlib.pyplot as plt
```

###CREATING DATASET

```python
x=np.array([[12],[23],[34],[45],[56],[67],[78],[89],[123],[134]])
y=np.array([240,1135,2568,4521,7865,9236,11932,14589,19856,23145])
```

###CONVERTING THE DATA INTO REQUIRED POLYNOMIAL FEATURES

```python
poly=PolynomialFeatures(degree=3)
X=poly.fit_transform(x)
```

###TRAINING THE MODEL

```python
model=LinearRegression()
model.fit(X,y)
```

###PREDICTING THE OUTPUT

```python
y_pred=model.predict(X)
print("coeff: ",model.coef_)
print("intercept: ",model.intercept_)
print("y predicted: ",y_pred)
```

###EVALUATING THE MODEL

```python
datalen=len(x)
r2=r2_score(y,y_pred)
print("r2 score=",r2)
sse=mean_squared_error(y,y_pred)*datalen
print("sse=",sse)
```

###DATA VISUALIZATION

```python
plt.scatter(x,y,label="data points",color="green")
plt.plot(x,y_pred,label="regression line",color="red")
plt.xlabel("x")
plt.ylabel("y predicted")
plt.title("polynomial regression")
plt.legend()
plt.show()
```
